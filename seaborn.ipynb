#1. Contagem de usuários por plano

```python
plt.figure(figsize=(8, 5))

ordem_planos = df_sistema['plano'].value_counts().index

sns.countplot(
    data=df_sistema,
    x='plano',
    order=ordem_planos
)

plt.title('Quantidade de Usuários por Plano')
plt.xlabel('Plano')
plt.ylabel('Quantidade de Usuários')
plt.tight_layout()
plt.show()
```

# Distribuição da idade com KDE

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df_sistema,
    x='idade',
    kde=True,
    bins=20
)

plt.title('Distribuição da Idade dos Usuários')
plt.xlabel('Idade')
plt.ylabel('Frequência')
plt.tight_layout()
plt.show()
```

# 3. Boxplot do tempo de uso por plano

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df_sistema,
    x='plano',
    y='tempo_uso_horas',
    order=['Gratuito', 'Básico', 'Pro']
)

plt.title('Distribuição do Tempo de Uso por Plano')
plt.xlabel('Plano')
plt.ylabel('Tempo de Uso (horas)')
plt.tight_layout()
plt.show()
```

# 4. Violinplot de satisfação por plano

```python
plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df_sistema,
    x='plano',
    y='satisfacao',
    order=['Gratuito', 'Básico', 'Pro']
)

plt.title('Distribuição da Satisfação por Plano')
plt.xlabel('Plano')
plt.ylabel('Satisfação')
plt.tight_layout()
plt.show()
```

# 5. Scatterplot: tempo de uso × satisfação

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_sistema,
    x='tempo_uso_horas',
    y='satisfacao',
    hue='plano',
    size='bugs_reportados',
    sizes=(30, 200),
    alpha=0.7
)

plt.title('Relação entre Tempo de Uso e Satisfação')
plt.xlabel('Tempo de Uso (horas)')
plt.ylabel('Satisfação')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

# Matriz de correlação de Pearson

```python
# Selecionando apenas as variáveis numéricas
df_numerico = df_sistema.select_dtypes(include='number')

# Calculando a correlação de Pearson
matriz_correlacao = df_numerico.corr(method='pearson')

# Visualizando a matriz
plt.figure(figsize=(8, 6))

sns.heatmap(
    matriz_correlacao,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0,
    linewidths=0.5
)

plt.title('Matriz de Correlação de Pearson')
plt.tight_layout()
plt.show()
```

#7. Bônus — Pairplot

```python
sns.pairplot(
    df_sistema,
    vars=[
        'tempo_uso_horas',
        'bugs_reportados',
        'satisfacao'
    ],
    hue='plano',
    diag_kind='hist',
    plot_kws={'alpha': 0.6}
)

plt.suptitle(
    'Pairplot — Uso, Bugs e Satisfação por Plano',
    y=1.02
)

plt.show()
```
