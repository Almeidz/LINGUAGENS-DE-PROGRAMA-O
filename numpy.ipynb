```python
import numpy as np
```

# Passo 1 — Array de peças vendidas

```python
pecas_vendidas = np.array([150, 120, 90, 210, 300, 250, 180])

print(pecas_vendidas)
```

# Passo 2 — Faturamento

```python
faturamento = pecas_vendidas * 50

print(faturamento)
```

# Passo 3 — Dias com mais de 200 peças

```python
mascara = pecas_vendidas > 200

print(mascara)
print(pecas_vendidas[mascara])
```

# Nível 2 — Múltiplas Filiais
Criar a matriz

```python
vendas = np.array([
    [200, 220, np.nan, 250],
    [150, 180, 160, 190],
    [300, 310, 290, 330]
])

print(vendas)
```

# Passo 2 — Total por loja

```python
total_por_loja = np.nansum(vendas, axis=1)

print(total_por_loja)
```

# Passo 3 — Média diária geral ignorando

```python
media_geral = np.nanmean(vendas)

print(media_geral)
```

# Passo 4 — Verificar se a meta foi atingida

```python
resultado_meta = np.where(
    vendas >= 200,
    'Meta Atingida',
    'Abaixo'
)

print(resultado_meta)
```

# Nível 3 — Avançado
Passo 1 — Gerar 50 vendas aleatórias

```python
rng = np.random.default_rng(42)

categorias = np.array(['Eletrônicos', 'Roupas', 'Casa'])

vendas_clientes = rng.choice(categorias, size=50)

print(vendas_clientes)
```

# Passo 2 — Contar vendas por departamento

```python
departamentos, quantidades = np.unique(
    vendas_clientes,
    return_counts=True
)

print(departamentos)
print(quantidades)
```

# Passo 3 — Encontrar a campanha com maior faturamento

```python
faturamento_campanhas = np.array([12000, 45000, 23000, 89000, 31000])

indice_campea = np.argmax(faturamento_campanhas)

print(indice_campea)
```

# Atividade 2 — Organização e Transformação de Dados
Nível 1 — Básico

```python
dias = np.arange(1, 31)

print(dias)
```

# Passo 2 — Metas igualmente espaçadas

```python
metas = np.linspace(20000, 30000, 5)

print(metas)
```

# Passo 3 — Converter estoque para inteiro

```python
estoque = np.array([10.5, 20.1, 30.9])

estoque_inteiro = estoque.astype(int)

print(estoque_inteiro)
```

# Passo 4 — Últimos 5 dias em ordem inversa

```python
ultimos_5 = dias[-5:][::-1]

print(ultimos_5)
```

# Nível 2 — Intermediário

```python
visitas = np.array([
    1000, 1200, 1500,
    1800, 2000, 2200,
    2500, 2700, 3000,
    3200, 3500, 3800
])

visitas_trimestres = visitas.reshape(4, 3)

print(visitas_trimestres)
```

# Passo 2 — vstack()

```python
primeiro_semestre = np.array([
    [1000, 1200, 1500],
    [1800, 2000, 2200]
])

segundo_semestre = np.array([
    [2500, 2700, 3000],
    [3200, 3500, 3800]
])

ano_completo = np.vstack([
    primeiro_semestre,
    segundo_semestre
])

print(ano_completo)
```

# Passo 3 — Copiar o primeiro trimestre

```python
primeiro_trimestre = ano_completo[0].copy()

print(primeiro_trimestre)
```

# Nível 3 — Avançado
Passo 1 — Ranking com argsort()

```python
pontuacoes = np.array([85, 92, 78, 95, 88])

indices_ranking = np.argsort(pontuacoes)

print(indices_ranking)
```

# Passo 2 — np.newaxis e Broadcasting

```python
precos = np.array([100, 200, 300, 400])

precos_coluna = precos[:, np.newaxis]

print(precos_coluna)
```

# Passo 3 — Sistema de equações com np.linalg.solve()

```python
A = np.array([
    [2, 1],
    [1, -1]
])

b = np.array([500, 100])

solucao = np.linalg.solve(A, b)

print("Custo da placa robótica: R$", solucao[0])
print("Custo do sensor: R$", solucao[1])
```
