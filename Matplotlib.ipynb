# 1. Preparação dos dados

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Gerando dados sintéticos estruturados
np.random.seed(42)
n_registros = 120

datas = pd.date_range(
    start='2024-01-01',
    periods=n_registros,
    freq='D'
)

categorias = ['Eletronicos', 'Livros', 'Roupas', 'Automotivo']
estados = ['SP', 'RJ', 'MG', 'RS', 'BA']

dados = {
    'data': datas,
    'categoria': np.random.choice(categorias, size=n_registros),
    'estado': np.random.choice(estados, size=n_registros),
    'valor': np.random.uniform(50, 4500, size=n_registros),
    'quantidade': np.random.randint(1, 15, size=n_registros),
}

df = pd.DataFrame(dados)

df['valor_venda'] = (
    df['valor'] *
    np.random.uniform(0.9, 1.1, size=n_registros)
).round(2)

df['data'] = pd.to_datetime(df['data'])
df = df.set_index('data')
```

# Faturamento mensal

```python
faturamento_mensal = df['valor_venda'].resample('ME').sum()

plt.figure(figsize=(10, 5))

plt.plot(
    faturamento_mensal.index,
    faturamento_mensal.values,
    marker='o',
    linestyle='--',
    color='royalblue'
)

plt.title('Faturamento Mensal')
plt.xlabel('Mês')
plt.ylabel('Faturamento (R$)')
plt.grid(True, linestyle=':', alpha=0.7)

plt.tight_layout()
plt.show()
```

# Total de vendas por categoria

```python
vendas_categoria = (
    df.groupby('categoria')['valor_venda']
    .sum()
    .sort_values()
)

plt.figure(figsize=(9, 5))

plt.bar(
    vendas_categoria.index,
    vendas_categoria.values,
    color='steelblue'
)

plt.title('Total de Vendas por Categoria')
plt.xlabel('Categoria')
plt.ylabel('Vendas (R$)')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

# Histograma dos valores das vendas

```python
plt.figure(figsize=(9, 5))

plt.hist(
    df['valor'],
    bins=10,
    edgecolor='black',
    color='skyblue'
)

plt.title('Distribuição dos Valores das Vendas')
plt.xlabel('Valor da Venda (R$)')
plt.ylabel('Frequência')

plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

# Subplots 2 × 1 com eixo X compartilhado

```python
quantidade_mensal = df['quantidade'].resample('ME').sum()

Depois criamos os dois gráficos:

fig, (ax1, ax2) = plt.subplots(
    2,
    1,
    figsize=(11, 8),
    sharex=True
)

# --------------------------------
# AX1 - Faturamento mensal
# --------------------------------

ax1.plot(
    faturamento_mensal.index,
    faturamento_mensal.values,
    marker='o',
    linestyle='--',
    color='royalblue'
)

ax1.set_title('Faturamento Mensal')
ax1.set_ylabel('Faturamento (R$)')
ax1.grid(True, linestyle=':', alpha=0.7)


# --------------------------------
# AX2 - Quantidade mensal
# --------------------------------

ax2.bar(
    quantidade_mensal.index,
    quantidade_mensal.values,
    color='seagreen'
)

ax2.set_title('Quantidade Total de Itens Vendidos por Mês')
ax2.set_xlabel('Mês')
ax2.set_ylabel('Quantidade')
ax2.grid(axis='y', linestyle=':', alpha=0.7)

plt.tight_layout()
plt.show()

sharex=True
```

# Gráfico com dois eixos Y usando twinx()

```python
quantidade_media_mensal = (
    df['quantidade']
    .resample('ME')
    .mean()
)
```

```python
fig, ax = plt.subplots(figsize=(11, 6))

# Eixo Y primário
ax.bar(
    faturamento_mensal.index,
    faturamento_mensal.values,
    color='cornflowerblue',
    alpha=0.75,
    label='Faturamento'
)

ax.set_xlabel('Mês')
ax.set_ylabel('Faturamento (R$)', color='royalblue')
ax.tick_params(axis='y', labelcolor='royalblue')


# Eixo Y secundário
ax2 = ax.twinx()

ax2.plot(
    quantidade_media_mensal.index,
    quantidade_media_mensal.values,
    color='darkorange',
    marker='o',
    linewidth=2,
    label='Quantidade média'
)

ax2.set_ylabel(
    'Quantidade Média de Itens por Venda',
    color='darkorange'
)

ax2.tick_params(
    axis='y',
    labelcolor='darkorange'
)

ax.set_title(
    'Faturamento Mensal x Quantidade Média de Itens por Venda'
)

ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

# Anotação do mês de maior faturamento

```python
mes_maior_faturamento = faturamento_mensal.idxmax()
maior_faturamento = faturamento_mensal.max()

print(
    f'Mês de maior faturamento: '
    f'{mes_maior_faturamento.strftime("%m/%Y")}'
)

print(
    f'Faturamento: R$ {maior_faturamento:,.2f}'
)
```

```python
fig, ax = plt.subplots(figsize=(11, 6))

ax.plot(
    faturamento_mensal.index,
    faturamento_mensal.values,
    marker='o',
    linestyle='--',
    color='royalblue'
)

# Identificando o pico
mes_maior_faturamento = faturamento_mensal.idxmax()
maior_faturamento = faturamento_mensal.max()

# Anotação
ax.annotate(
    f'Maior faturamento\nR$ {maior_faturamento:,.2f}',
    xy=(mes_maior_faturamento, maior_faturamento),
    xytext=(20, 30),
    textcoords='offset points',
    arrowprops=dict(
        arrowstyle='->',
        color='red',
        linewidth=2
    ),
    fontsize=10,
    color='darkred',
    bbox=dict(
        boxstyle='round,pad=0.4',
        facecolor='white',
        edgecolor='red'
    )
)

ax.set_title('Faturamento Mensal com Destaque para o Pico')
ax.set_xlabel('Mês')
ax.set_ylabel('Faturamento (R$)')
ax.grid(True, linestyle=':', alpha=0.7)

plt.tight_layout()
plt.show()
```

# Criando a tabela dinâmica

```python
tabela_pivot = pd.pivot_table(
    df,
    values='valor_venda',
    index='categoria',
    columns='estado',
    aggfunc='sum'
)

print(tabela_pivot)
```

# Heatmap utilizando imshow()

```python
plt.figure(figsize=(10, 6))

plt.imshow(
    tabela_pivot,
    cmap='YlOrRd',
    aspect='auto'
)

# Barra de cores
plt.colorbar(label='Valor das Vendas (R$)')

# Rótulos do eixo X
plt.xticks(
    ticks=np.arange(len(tabela_pivot.columns)),
    labels=tabela_pivot.columns
)

# Rótulos do eixo Y
plt.yticks(
    ticks=np.arange(len(tabela_pivot.index)),
    labels=tabela_pivot.index
)

plt.title('Heatmap de Vendas por Categoria e Estado')
plt.xlabel('Estado')
plt.ylabel('Categoria')

plt.tight_layout()
plt.show()
```

# Salvando o Heatmap em PNG

```python
plt.figure(figsize=(10, 6))

plt.imshow(
    tabela_pivot,
    cmap='YlOrRd',
    aspect='auto'
)

plt.colorbar(label='Valor das Vendas (R$)')

plt.xticks(
    ticks=np.arange(len(tabela_pivot.columns)),
    labels=tabela_pivot.columns
)

plt.yticks(
    ticks=np.arange(len(tabela_pivot.index)),
    labels=tabela_pivot.index
)

plt.title('Heatmap de Vendas por Categoria e Estado')
plt.xlabel('Estado')
plt.ylabel('Categoria')

plt.tight_layout()

plt.savefig(
    'heatmap_vendas.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```
